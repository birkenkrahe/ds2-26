## Relational Operators
1. Create a vector `ages` with values `18, 25, 30, 17, 40`[cite: 5]. Write code to find which ages are greater than or equal to 21, and print these `ages`[cite: 6].

```r
ages <- c(18, 25, 30, 17, 40)
greater_than_21 <- ages >= 21
print(greater_than_21)
print(ages[greater_than_21])
```

3. Compare two strings, `"data"` and `"science"`, and determine which comes first alphabetically[cite: 8]. Use `cat` to reproduce the output: `"data" comes before "science": true`[cite: 9].

```r
first <- "data"
second <- "science"
cat('"', first, '" comes before "', second, '": ', tolower(first < second), sep="")
```

## Logical Operators
1. Create a vector `scores <- c(55, 70, 75, 45, 80)`[cite: 9, 10]. Use logical operators to check which scores are in the open interval (50, 75)[cite: 10]. Store the result in a logical index vector `idx` and use it to print the values[cite: 11, 12].

```r
scores <- c(55, 70, 75, 45, 80)
idx <- scores > 50 & scores < 75
scores[idx]
```

2. Write a script to check if a number `num <- 15` is divisible by both 3 and 5. Print the result as a Boolean value[cite: 12, 13].

```r
num <- 15
(num %% 3 == 0) & (num %% 5 == 0)
```

3. Create two vectors, `a <- c(TRUE, FALSE, TRUE)` and `b <- c(FALSE, TRUE, TRUE)`[cite: 13]. Use logical operators to find the results of the operations: `a AND b` and `a OR b`[cite: 14, 15].

```r
a <- c(TRUE, FALSE, TRUE)
b <- c(FALSE, TRUE, TRUE)
print(a & b)
print(a | b)
```

## Conditional Statements
1. Create a variable `temperature` with the value `25`[cite: 15]. Write a conditional statement to print `"cold"` if temperature is below 20, and `"Warm"` otherwise[cite: 16, 17].

```r
temperature <- 25
if (temperature < 20) {
    print("cold")
} else {
    print("Warm")
}
```

2. Write a script that checks if a given number `x` is negative, zero, or positive, and prints the appropriate message. Test for 5, -10, 0[cite: 17].

```r
check_val <- function(x) {
    if (x > 0) {
        print("positive")
    } else if (x < 0) {
        print("negative")
    } else {
        print("zero")
    }
}

check_val(5)
check_val(-10)
check_val(0)
```

3. Write a conditional statement to check if a given vector of numbers `foo` contains any value greater than 100[cite: 17]. If it does, print `"outlier detected."`, otherwise, print `"all values within range."`[cite: 18].

```r
foo <- c(10, 50, 101, 20)
if (any(foo > 100)) {
    print("outlier detected.")
} else {
    print("all values within range.")
}
```
